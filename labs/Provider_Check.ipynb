# ✅ Provider Check (2 minutes)

This tiny notebook confirms your API setup works before you start any lab.
It runs one chat call and one embeddings call and prints a green/red summary.

**Default (OpenAI):** add the `OPENAI_API_KEY` Colab Secret (key icon, enable
notebook access) and Run All.

**Alternative provider (for students who cannot use OpenAI):** add these
Colab Secrets instead, then Run All. See `labs/ALTERNATIVE_PROVIDERS.md`
for how to create the accounts.

| Secret | Value |
|---|---|
| `LLM_PROVIDER` | `deepseek` |
| `DEEPSEEK_API_KEY` | your DeepSeek key |
| `QWEN_API_KEY` | your Qwen key (only needed for the embeddings labs M01/M05/M10) |

**Jupyter instead of Colab?** Set the same names as environment variables
first, or paste the key at the hidden prompt when the cell asks.


```python
# === Setup: install the course utils and connect ===
!pip install -q dads5250==0.2.0

from dads5250 import pp, setup_openai, DEFAULT_CHAT_MODEL, DEFAULT_MINI_MODEL, DEFAULT_EMBED_MODEL

client = setup_openai()   # loads your key, verifies the connection, prints the provider
```

```python
# === The checks: one chat call, one embeddings call ===
results = {}

# chat: every lab in the course needs this one
try:
    r = client.chat.completions.create(
        model=DEFAULT_MINI_MODEL,
        temperature=0,
        messages=[{"role": "user", "content": "Reply with exactly: chat works"}],
    )
    results["chat"] = "PASS: " + r.choices[0].message.content.strip()
except Exception as e:
    results["chat"] = f"FAIL: {e}"

# embeddings: needed for labs M01 Lab 2, M05 and M10 only
try:
    e = client.embeddings.create(model=DEFAULT_EMBED_MODEL, input="hello world")
    results["embeddings"] = f"PASS: vector of {len(e.data[0].embedding)} numbers"
except Exception as ex:
    results["embeddings"] = (f"not available: {ex}  "
                             "(chat labs still work; add QWEN_API_KEY for the embeddings labs)")

results["chat model"] = DEFAULT_MINI_MODEL
results["reasoning model"] = DEFAULT_CHAT_MODEL
pp(results, title="Provider check summary")
```

**What a good result looks like:** `chat` says PASS and shows `chat works`;
`embeddings` says PASS with a vector length. If `chat` fails, re-check the
Secret names and that notebook access is enabled on each Secret. If only
`embeddings` fails on an alternative provider, add the `QWEN_API_KEY` Secret.
